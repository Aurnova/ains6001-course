```python
import numpy as np

rng = np.random.default_rng(36)
n = 240
truth = rng.binomial(1, 0.45, n)
difficulty = rng.uniform(0, 1, n)
ai_score = np.clip(0.16 + 0.68*truth + rng.normal(0, 0.12 + 0.25*difficulty), 0, 1)
ai_decision = ai_score >= 0.5
human_correct = rng.random(n) < (0.90 - 0.28*difficulty)
human_decision = np.where(human_correct, truth, 1-truth)
uncertain = np.abs(ai_score - 0.5) < 0.16
joint_decision = np.where(uncertain, human_decision, ai_decision)
human_accuracy = float((human_decision == truth).mean())
ai_accuracy = float((ai_decision == truth).mean())
joint_accuracy = float((joint_decision == truth).mean())
review_workload = float(uncertain.mean())
print({"human_accuracy": round(human_accuracy, 3), "ai_accuracy": round(ai_accuracy, 3),
       "joint_accuracy": round(joint_accuracy, 3), "review_workload": round(review_workload, 3)})
```
